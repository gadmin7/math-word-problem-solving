```python
import os
import torch
import numpy as np
import pandas as pd
from transformers import (
    AutoModelForCausalLM,
    AutoTokenizer,
    BitsAndBytesConfig,
    DataCollatorForLanguageModeling,
    TrainingArguments,
    AutoConfig
)
from datasets import load_dataset, concatenate_datasets
from huggingface_hub import login
from peft import LoraConfig, get_peft_model,PeftModel
from trl import SFTTrainer
import transformers
```

```python
print(f"Number of GPUs Available:{torch.cuda.device_count()}")
```

```python
token = ""
login(token)
```

```python
save_dir = "/kaggle/working"
model_id = "mistralai/Mistral-7B-Instruct-v0.1"
```

```python
lora_config = LoraConfig(
    r=8,
    lora_alpha=16,
    lora_dropout=0.1,
    target_modules=[
        "q_proj", "k_proj", "v_proj", "o_proj",
        "gate_proj", "up_proj", "down_proj"
    ],
    bias="none",
    task_type="CAUSAL_LM",
)
```

```python
bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_use_double_quant=True,
    bnb_4bit_quant_type='nf4',
    bnb_4bit_compute_dtype=torch.float16
)
```

```python
tokenizer = AutoTokenizer.from_pretrained(model_id, token=token)
```

```python
# Define default tokens
DEFAULT_PAD_TOKEN = "[PAD]"
DEFAULT_EOS_TOKEN = "</s>"
DEFAULT_BOS_TOKEN = "<s>"

special_tokens_dict = {}
if tokenizer.pad_token is None:
    special_tokens_dict['pad_token'] = DEFAULT_PAD_TOKEN
if tokenizer.eos_token is None:
    special_tokens_dict['eos_token'] = DEFAULT_EOS_TOKEN
if tokenizer.bos_token is None:
    special_tokens_dict['bos_token'] = DEFAULT_BOS_TOKEN

if special_tokens_dict:
    tokenizer.add_special_tokens(special_tokens_dict)
```

```python
model = AutoModelForCausalLM.from_pretrained(
    model_id,
    quantization_config=bnb_config,
    device_map="auto",
    token=token
)
```

```python
# Resize the model's token embeddings after adding new tokens
model.resize_token_embeddings(len(tokenizer))

# Apply LoRA to the model
model = get_peft_model(model, lora_config)

# Set tokenizer padding side
tokenizer.padding_side = "right"
```

```python
def preprocess_function(examples):
    # Format inputs with the instruction
    inputs = [
        f"[INST] Solve the following math problem: {problem}\n"
        "Show all intermediate steps and include the final answer in LaTeX format in a box like \\boxed{{}}.[/INST]"
        for problem in examples['problem']
    ]
    # Append the solution with EOS token
    targets = [
        f"{solution}{tokenizer.eos_token}"
        for solution in examples['solution']
    ]
    # Concatenate inputs and targets
    full_texts = [inp + tgt for inp, tgt in zip(inputs, targets)]
    # Tokenize the concatenated texts
    model_inputs = tokenizer(
        full_texts,
        max_length=512,
        truncation=True,
        padding="longest",  # Use dynamic padding
        return_tensors="pt"
    )
    # Create labels by cloning input IDs
    labels = model_inputs["input_ids"].clone()

    # Mask input tokens in labels
    for i in range(len(labels)):
        input_ids = tokenizer(inputs[i], add_special_tokens=False).input_ids
        input_len = len(input_ids)
        labels[i][:input_len] = -100  # Mask the input tokens
    model_inputs["labels"] = labels
    return model_inputs
```

```python
data_collator = DataCollatorForLanguageModeling(
    tokenizer=tokenizer,
    mlm=False,  # For causal language modeling
    pad_to_multiple_of=8  # For faster GPU performance
)
```

```python
dataset = load_dataset("lighteval/MATH", "algebra")
```

```python
for level in range(1, 5):
    print(f"\nStarting training for Level {level}")

    if level == 1:
       
        model = AutoModelForCausalLM.from_pretrained(
            model_id,
            quantization_config=bnb_config,
            device_map="auto",
            token=token,
        )
        model.resize_token_embeddings(len(tokenizer))
        model = get_peft_model(model, lora_config)
        tokenizer.padding_side = "right"
    else:
        # Load tokenizer and model from previous level
        prev_level = level - 1
        prev_model_path = os.path.join(save_dir, f"model_level_{prev_level}")
        
        model = PeftModel.from_pretrained(
        model=model,
        model_id = prev_model_path,
        peft_config=bnb_config,
        device_map="auto")
        
        model.resize_token_embeddings(len(tokenizer))
        model = get_peft_model(model, lora_config)
        tokenizer.padding_side = "right"
    
    # Filter the dataset by current level
    level_train = dataset['train'].filter(
        lambda x: x['level'] == f"Level {level}"
    )
    level_test = dataset['test'].filter(
        lambda x: x['level'] <= f"Level {level}"
    )
    
    print(f"Number of training samples: {len(level_train)}")
    print(f"Number of test samples: {len(level_test)}")
    
    # Combine with previous levels' data if applicable
    if level > 1:
        replay_fraction = 0.1  # Adjust as needed
        previous_levels_train = dataset['train'].filter(
            lambda x: int(x['level'].split()[-1]) < level
        ).shuffle(seed=42).select(range(int(len(level_train)*replay_fraction)))
        level_train = concatenate_datasets([level_train, previous_levels_train])
    
    # Tokenize datasets
    level_train = level_train.map(preprocess_function, batched=True)
    level_test = level_test.map(preprocess_function, batched=True)
    
    training_args = TrainingArguments(
      output_dir=os.path.join(save_dir, f"results_level_overall"),
      per_device_train_batch_size=1,
      gradient_accumulation_steps=4,
      warmup_steps=100,
      num_train_epochs=2,
      learning_rate=1e-4,
      fp16=True,  # Use bf16 precision
      logging_steps=10,
      optim="paged_adamw_8bit",
      evaluation_strategy="steps",
      eval_steps=100,
      save_steps=100,
      save_total_limit=2,
      report_to="none",  # Change to "wandb" if using Weights & Biases
      run_name=f"Mistral_FineTuning_Level_overall",
        ddp_find_unused_parameters=False,
    )
    
    # Initialize the trainer
    trainer = SFTTrainer(
        model=model,
        train_dataset=level_train,
        eval_dataset=level_test,
        peft_config=lora_config,
        args=training_args,
        data_collator=data_collator,
    )
    
    # Train the model
    trainer.train()
    
    # Save the model
    model_save_path = os.path.join(save_dir, f"model_level_{level}")
    model.save_pretrained(model_save_path,save_embedding_layers=True)
    tokenizer.save_pretrained(model_save_path)
    
    # Evaluate the model
    results = trainer.evaluate()
    print(f"Overall evaluation results: {results}")

    # Generate predictions on the test set
    model.eval()  # Set model to evaluation mode
    results_list = []
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    for idx in range(len(level_test)):
        sample = level_test[idx]
          # Reconstruct the input prompt
        input_text = (
              f"[INST] Solve the following math problem: {sample['problem']}\n"
              "Show all intermediate steps and include the final answer in LaTeX format in a box like \\boxed{{}}.[/INST]"
          )
        ground_truth = sample['solution']

          # Tokenize the input text
        input_ids = tokenizer.encode(input_text, return_tensors='pt').to(model.device)

          # Generate prediction with no_grad to save memory
        with torch.no_grad():
            output_ids = model.generate(
                  input_ids,
                  max_length=1024,
                  repetition_penalty=1.2,
                  early_stopping=True
              )
        try:
            predicted_text = tokenizer.decode(output_ids[0], skip_special_tokens=True)
        except Exception as e:
            predicted_text = e

          # Store the results
        results_list.append({
              "problem": sample['problem'],
              "level":sample['level'],
              "ground_truth": ground_truth,
              "predicted_solution": predicted_text
          })
        if (idx + 1) % 100 == 0:
            results_df = pd.DataFrame(results_list)
            results_save_path = os.path.join(save_dir, f"test_results_level_{level}_{idx+1}.csv")
            results_df.to_csv(results_save_path, index=False)
            print(f"Saved test results up to sample {idx+1} to {results_save_path}")

    # Save the test results to a CSV file
    results_df = pd.DataFrame(results_list)
    results_save_path = os.path.join(save_dir, f"test_results_level_{level}.csv")
    results_df.to_csv(results_save_path, index=False)
    print(f"Saved test results to {results_save_path}")
```
